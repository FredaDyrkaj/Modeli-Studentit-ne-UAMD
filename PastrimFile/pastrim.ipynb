# Pastrim i Tot_Students_FTI_BUSS_FSHPJ_Rikodifikuar

Bashkon skemat FTI/FSHPJ (shqip) dhe FB (anglisht), mbyll rikodifikimin e mbetur, validon notat 4–10 dhe ruan `Studentet_Uamd_pastruar.xlsx`.

```python
from pathlib import Path

import pandas as pd

from student_data import (
    INPUT_TOT,
    numri_bosh,
    pastro_dataframe,
    shtyp_qa,
)

INPUT_XLSX = Path(INPUT_TOT)
OUTPUT_XLSX = Path("Studentet_Uamd_pastruar.xlsx")
```

```python
# 1) Leximi i skedarit të rikodifikuar
Studentet_Uamd_raw = pd.read_excel(INPUT_XLSX, sheet_name=0, engine="openpyxl")
print("Forma (rreshta, kolona):", Studentet_Uamd_raw.shape)
Studentet_Uamd_raw.head()
```

```python
# 2) Boshllëqe përpara pastrimit (NaN + string bosh)
bosh_per_kolon = pd.Series(
    {c: numri_bosh(Studentet_Uamd_raw[c]) for c in Studentet_Uamd_raw.columns}
).sort_values(ascending=False)
print("Boshllëqe sipas kolonës:")
print(bosh_per_kolon.to_frame("rreshta_bosh").to_string())
```

```python
Studentet_Uamd_raw.info()
```

```python
# 3) Pastrim: bashkim skemash, rikodifikim i mbetur, nota, data, deduplikim
# Nuk hiqen rreshtat vetëm sepse Baze_/Zgjedhje_ janë bosh (FB ka Base_*).
# Dublikatat hiqen vetëm kur ID_STUDENTI është i mbushur; Bachelor+Master i njëjti ID ruhen.
Studentet_Uamd = pastro_dataframe(Studentet_Uamd_raw)
print("Forma pas pastrimit:", Studentet_Uamd.shape)
Studentet_Uamd.head()
```

```python
# 4) QA
shtyp_qa(Studentet_Uamd)
```

```python
# 5) Ruajtja
out_path = Path(OUTPUT_XLSX)
out_path.parent.mkdir(parents=True, exist_ok=True)
Studentet_Uamd.to_excel(out_path, index=False, engine="openpyxl")
print("U shkrua:", out_path.resolve())
```
